## Evaluate Model on Validation Data

Visualize model predictions on real validation data to check detection quality.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import torch
from scipy.ndimage import label, center_of_mass, gaussian_filter
from scipy.spatial.distance import cdist
from scipy.optimize import linear_sum_assignment

from modules.utils import open_tiff_file

# Load validation data
video_stack = open_tiff_file('/content/val_data/val.tif').astype(np.float32)
gt_df = pd.read_csv('/content/val_data/val.csv')

# ROI settings
roi_y_min, roi_y_max = 512, 768
roi_x_min, roi_x_max = 256, 512

# Detection parameters
THRESHOLD = 0.5
MIN_AREA = 4
SIGMA = 1.0

def extract_detections(prob_map, threshold=0.5, min_area=4, sigma=1.0):
    if sigma > 0:
        prob_map = gaussian_filter(prob_map, sigma=sigma)
    binary = (prob_map > threshold).astype(np.uint8)
    labeled, num = label(binary)
    detections = []
    for i in range(1, num + 1):
        mask = (labeled == i)
        if mask.sum() < min_area:
            continue
        y, x = center_of_mass(mask)
        if not np.isnan(x) and not np.isnan(y):
            detections.append((x, y))
    return np.array(detections)

def calculate_deta(gt, pred, thresh=5.0):
    if len(gt) == 0 and len(pred) == 0: return 1.0
    if len(gt) == 0 or len(pred) == 0: return 0.0
    dists = cdist(gt, pred)
    sims = np.maximum(0, 1 - dists / thresh)
    row, col = linear_sum_assignment(-sims)
    tp = sum(sims[r, c] > 0 for r, c in zip(row, col))
    return tp / (len(gt) + len(pred) - tp)

# Visualize on sample frames
frames_to_show = [0, 10, 25, 40, 50]
fig, axes = plt.subplots(len(frames_to_show), 3, figsize=(14, 4 * len(frames_to_show)))

print(f"\nEvaluating on {len(frames_to_show)} sample frames...")
print(f"Detection params: threshold={THRESHOLD}, min_area={MIN_AREA}, sigma={SIGMA}\n")

with torch.no_grad():
    for i, frame_idx in enumerate(frames_to_show):
        # Crop & normalize
        crop = video_stack[frame_idx, roi_y_min:roi_y_max, roi_x_min:roi_x_max]
        crop_norm = (crop - crop.mean()) / (crop.std() + 1e-8)

        # Inference
        tensor = torch.from_numpy(crop_norm).unsqueeze(0).unsqueeze(0).float().to(DEVICE)
        prob_map = torch.sigmoid(model(tensor)).cpu().numpy()[0, 0]

        # Extract detections
        pred = extract_detections(prob_map, THRESHOLD, MIN_AREA, SIGMA)

        # Ground truth
        gt_sub = gt_df[(gt_df['frame'] == frame_idx) &
                       gt_df['x'].between(roi_x_min, roi_x_max) &
                       gt_df['y'].between(roi_y_min, roi_y_max)]
        gt = gt_sub[['x', 'y']].values - [roi_x_min, roi_y_min]

        # Calculate DetA
        deta = calculate_deta(gt, pred)

        # Plot input
        axes[i, 0].imshow(crop, cmap='gray')
        axes[i, 0].set_title(f"Frame {frame_idx} - Input")
        axes[i, 0].axis('off')

        # Plot probability map
        axes[i, 1].imshow(prob_map, cmap='hot', vmin=0, vmax=1)
        axes[i, 1].set_title(f"Probability Map")
        axes[i, 1].axis('off')

        # Plot detections
        axes[i, 2].imshow(crop, cmap='gray')
        if len(gt) > 0:
            axes[i, 2].scatter(gt[:, 0], gt[:, 1], s=120, facecolors='none',
                               edgecolors='lime', linewidth=2, label=f'GT ({len(gt)})')
        if len(pred) > 0:
            axes[i, 2].scatter(pred[:, 0], pred[:, 1], c='red', marker='x',
                               s=60, linewidth=2, label=f'Pred ({len(pred)})')
        axes[i, 2].legend(loc='upper right')
        axes[i, 2].set_title(f"Detections (DetA={deta:.3f})")
        axes[i, 2].axis('off')

        print(f"Frame {frame_idx}: GT={len(gt)}, Pred={len(pred)}, DetA={deta:.3f}")

plt.tight_layout()
plt.savefig(os.path.join(CONFIG['SAVE_DIR'], 'detection_evaluation.png'), dpi=150)
plt.show()

print(f"\n✓ Evaluation visualization saved")
```

## Run LapTrack Tracking with Custom Parameters

- Run tracking on validation data using LapTrack algorithm with specified parameters.

- We have already enter the best parametres from the sweep for this setup.

```python
from modules.eval.tracking import DetectionParams, LapTrackParams

# Detection parameters
DETECTION_PARAMS = DetectionParams(
    threshold=0.5,        # Detection threshold (0.0-1.0)
    min_area=4,            # Minimum detection area (pixels)
    nms_min_dist=3         # Non-maximum suppression distance
)

# LapTrack parameters
LAPTRACK_PARAMS = LapTrackParams(
    max_distance=6.0,              # Max linking distance between frames
    track_cost_cutoff=10.0,        # Cost cutoff for linking
    gap_closing_max_frame_count=3, # Max frames to close gaps
    gap_closing_cost_cutoff=15.0,  # Cost cutoff for gap closing
    min_track_len=5                # Minimum track length to keep
)

# ROI coordinates (Region of Interest)
Y_MIN, Y_MAX = 512, 768
X_MIN, X_MAX = 256, 512

print("Parameters configured:")
print(f"  Detection: threshold={DETECTION_PARAMS.threshold}, min_area={DETECTION_PARAMS.min_area}")
print(f"  LapTrack: max_distance={LAPTRACK_PARAMS.max_distance}, gap_closing={LAPTRACK_PARAMS.gap_closing_max_frame_count}")
```

```python
# @title Run LapTrack Tracking
from IPython.display import Image, display
from modules.eval.tracking import CCPDetector, track_with_laptrack, hota
from modules.eval.sweep import save_tracking_gif
from modules.utils import open_tiff_file
from modules.config import DEVICE
from tqdm.auto import tqdm

# Output path
GIF_OUTPUT_PATH = os.path.join(CONFIG['SAVE_DIR'], "laptrack_result.gif")

print("="*70)
print("LAPTRACK TRACKING")
print("="*70)

# Load validation data
val_tif_path = "/content/val_data/val.tif"
val_csv_path = "/content/val_data/val.csv"

print("\n[1/4] Loading validation data...")
val_input = open_tiff_file(val_tif_path).astype(np.float64)
val_gt = pd.read_csv(val_csv_path)

# Filter GT to ROI
val_gt = val_gt.groupby('track_id').filter(
    lambda t: (Y_MIN < t.y.mean() < Y_MAX) and (X_MIN < t.x.mean() < X_MAX)
)
print(f"  ✓ Loaded {len(val_input)} frames, {len(val_gt['track_id'].unique())} GT tracks in ROI")

# Crop to ROI
val_roi = val_input[:, Y_MIN:Y_MAX, X_MIN:X_MAX]

# Initialize detector
print("\n[2/4] Running detection...")
detector = CCPDetector(model, device=DEVICE, params=DETECTION_PARAMS)

detections_per_frame = []
for frame_idx in tqdm(range(len(val_roi)), desc="Detecting"):
    frame = val_roi[frame_idx]
    frame_norm = (frame - frame.mean()) / (frame.std() + 1e-8)
    _, detections = detector.detect(frame_norm)
    detections_per_frame.append(detections)

total_detections = sum(len(d) for d in detections_per_frame)
print(f"  ✓ Total detections: {total_detections}")

# Run LapTrack
print("\n[3/4] Running LapTrack...")
tracks_df = track_with_laptrack(detections_per_frame, LAPTRACK_PARAMS)

# Adjust coordinates to full image space
if not tracks_df.empty:
    tracks_df['x'] += X_MIN
    tracks_df['y'] += Y_MIN

num_tracks = tracks_df['track_id'].nunique() if not tracks_df.empty else 0
print(f"  ✓ Generated {num_tracks} tracks")

# Evaluate with HOTA
print("\n[4/4] Evaluating...")
if not tracks_df.empty:
    results = hota(val_gt, tracks_df)
    print(f"\n{'='*70}")
    print("RESULTS")
    print(f"{'='*70}")
    print(f"  HOTA: {results['HOTA']:.4f}")
    print(f"  DetA: {results['DetA']:.4f}")
    print(f"  AssA: {results['AssA']:.4f}")
    print(f"{'='*70}")
else:
    print("  ⚠ No tracks generated!")
    results = {}

# Generate and save GIF
print(f"\nGenerating GIF: {GIF_OUTPUT_PATH}")
save_tracking_gif(
    tracks_df,
    val_input,
    output_path=GIF_OUTPUT_PATH,
    y_min=Y_MIN, y_max=Y_MAX,
    x_min=X_MIN, x_max=X_MAX,
    tail_length=10,
    color='yellow',
    fps=10
)

print(f"\n✓ GIF saved to: {GIF_OUTPUT_PATH}")
```
